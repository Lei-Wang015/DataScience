```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#绘制图形
#横坐标：等差数列
x = np.linspace(0, 2 * np.pi, 100)
#纵坐标：正弦波
y = np.sin(x)
#画线型图
plt.plot(x, y)
```

```python
#全称x limits,设置横坐标范围是-1 ~ 10
plt.xlim(-1, 10)
#全称y limits,设置纵坐标范围是-4.5 ~ 1.5
plt.ylim(-1.5, 1.5)
#设置网格线，颜色为绿色，透明度0.5，显得央视为虚线，宽度为2
plt.grid(color = 'green', alpha = 0.5, linestyle = '--', linewidth = 2)
plt.plot(x, y)
```

```python
#设置标题
x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)
plt.plot(x, y)
#图的标题设置为Sin(x),字体大小为18,颜色为红色,标题与图片的间距为20
plt.title('Sin(x)', fontsize = 18, color = 'red', pad = 20) 
```

```python
x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)
plt.plot(x, y)
#matplotlib默认的字体库通常不包含中文字体，直接绘制中文时，会出现方框乱码（□□□）
#遂设置字体为楷体
#plt.rcParams是 matplotlib 的全局配置字典，用来控制绘图的各种默认样式。
#'font.family'：指定全局使用的字体家族。
#'KaiTi'：是 “楷体” 的字体名称（Windows 系统下的标准名称）。
plt.rcParams['font.family'] = 'KaiTi'
#图的标题设置为正弦波,字体大小为18,颜色为红色,标题与图片的间距为20
plt.title('正弦波', fontsize = 18, color = 'red', pad = 20) 
```

```python
x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)
plt.plot(x, y)
#设置字体为楷体
plt.rcParams['font.family'] = 'KaiTi'
#设置数字的负号
#axes.unicode_minus：控制 matplotlib 是否使用 Unicode 减号来显示负号。
#True（默认）：使用 Unicode 减号。
#False：关闭 Unicode 减号，改用 ASCII 减号（就是键盘上 - 这个字符）。
#这行代码是中文字体设置的 “配套补丁”，专门解决负号乱码问题
plt.rcParams['axes.unicode_minus'] = False
#图的标题设置为正弦波,字体大小为18,颜色为红色,标题与图片的间距为20
plt.title('正弦波', fontsize = 18, color = 'red', pad = 20) 
```

```python
#设置标签
plt.figure(figsize = (9, 6))
x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)
plt.plot(x, y)
plt.rcParams['font.family'] = 'KaiTi'
plt.rcParams['axes.unicode_minus'] = False
plt.rcParams['font.size'] = 28
plt.title('正弦波', fontsize = 18, color = 'red', pad = 20)

#设置横坐标为X
plt.xlabel('X')
#设置纵坐标为f(x) = sin(x),并把它横过来
#plt.ylabel() 的标签默认是垂直显示的（文字沿着 y 轴方向，竖排），对应的 rotation 默认值是 90 度。
#rotation = 0：将标签的旋转角度设置为 0 度，让它水平显示
plt.ylabel('f(x) = sin(x)', rotation = 0, horizontalalignment = 'right')
#horizontalalignment（也可简写为 ha）是 matplotlib 中用于控制文本在水平方向上的对齐方式的参数，它决定了文本相对于其锚点（位置坐标）如何排列。
#'center'（默认值）：文本以锚点为中心，水平居中对齐。
#'left'：文本的左边缘与锚点对齐，文本向右展开。
#'right'：文本的右边缘与锚点对齐，文本向左展开。
```
