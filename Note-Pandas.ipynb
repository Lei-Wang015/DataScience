```python
#concat数据串联
# df1 看做是一班的考试成绩
df1 = pd.DataFrame(data = np.random.randint(0, 150, size = (10, 3)),# 计算机科目的考试成绩
                  index = list('ABCDEFGHIJ'),  # 行标签，用户
                  columns = ['Python', 'Tensorflow', 'Keras']) # 考试科目

# df2 看做是二班的考试成绩
df2 = pd.DataFrame(data = np.random.randint(0, 150, size = (10,3)),# 计算机科目的考试成绩
                  index = list('KLMNOPQRST'),# 行标签，用户
                  columns = ['Python', 'Tensorflow', 'Keras']) # 考试科目

# df3 新增两门考试科目（一班）
df3 = pd.DataFrame(data = np.random.randint(0, 150, size = (10,2)),
                  index = list('ABCDEFGHIJ'),
                  columns = ['PyTorch', 'Paddle'])

display(df1, df2, df3)

# 一班二班考试成绩合并
# axis = 0 表示进行(xing)行(hang)合并, 即行增加
display(pd.concat([df1, df2], axis = 0))

# 一班科目增加，合并科目
# axis = 1 表示进行列合并，即列增加
display(pd.concat([df1, df3], axis = 1))
```

```python
#Join SQL风格合并
#记录name weight
df1 = pd.DataFrame(data = {'name':['apple', 'banana', 'orange', 'pineapple'],
                           'weight':[65, 60, 70, 150]})
#记录name height
df2 = pd.DataFrame(data = {'name':['apple', 'banana', 'orange', 'pineapple'],
                           'height':[176, 184, 178, 166]})
#记录fruit height
df3 = pd.DataFrame(data = {'fruit':['apple', 'banana', 'orange', 'pineapple'],
                           'height':[176, 184, 178, 166]})
display(df1, df2, df3)
#用merge对数据进行合并,df1 df3不能直接合并因为没有公共属性
display(pd.merge(df1, df2))
display(pd.merge(df2, df3, on = 'height'))
#对df1和df3进行指定合并
display(pd.merge(df1, df3, left_on = 'name', right_on = 'fruit')) #merge在默认情况下会保留所有列
```

```python
#用unstack实现多层索引的数据重塑
df = pd.DataFrame(data = np.random.randint(0, 100, size = (22, 3)),
                  index = pd.MultiIndex.from_product([list('ABCDEFGHIJK'), 
                                                      ['mid', 'final']]), 
                  columns = ['python', 'Tensorflow', 'Keras'])
display(df.unstack()) #把行索引最里层的期中期末 移到了列索引的位置
```

```python
#用unstack实现多层索引的数据重塑
df = pd.DataFrame(data = np.random.randint(0, 100, size = (22, 3)),
                  index = pd.MultiIndex.from_product([list('ABCDEFGHIJK'), 
                                                      ['mid', 'final']]), 
                  columns = ['python', 'Tensorflow', 'Keras'])
display(df.unstack(level = 0)) #默认level = 1
#从外到内计数：最外层（第一层）行索引是level=0，往里一层是level=1，再内层是level=2……
#从内到外计数（负数）：最内层是level=-1（默认值），倒数第二层是level=-2……
#level=0就是把最外层的行索引转成列索引；默认level=-1（等价于level=1）
```

```python
#简单统计指标
#创造数据
df = pd.DataFrame(data = np.random.randint(0, 100,size = (20, 3)),
                  index = list('ABCDEFGHIJKLMNOPQRST'),
                  columns = ['Python', 'Tensorflow', 'Keras'])
display(df)
#把部分数据设置为空数据
def convert(x):
    if x > 80:
        return np.nan
    else:
        return x
df['Python'] = df['Python'].map(convert)
df['Tensorflow'] = df['Tensorflow'].apply(convert)
df['Keras'] = df['Keras'].transform(convert)
display(df)
#用count函数统计空数据的个数
display(df.count())
```

```python
#更多统计指标
df = pd.DataFrame(data = np.random.randint(0, 100,size = (20, 3)),
                  index = np.arange(20),
                  columns = ['Python', 'Tensorflow', 'Keras'])
display(df)
#统计元素出现次数
display(df['Python'].value_counts())
#去重
display(df['Python'].unique())
#累加
display(df.cumsum())
#累乘
display(df.cumprod())
#累计最小值，即碰到更小的数后，该数后的全部数都变成这个更小的数
display(df.cummin())
#累计最大值
display(df.cummax())
#计算标准差
display(df.std())
#计算方差
display(df.var())
#计算差分
display(df.diff())
#计算百分比变化
display(df.pct_change())
```

```python
#高级统计指标
df = pd.DataFrame(data = np.random.randint(0, 100,size = (20, 3)),
                  index = np.arange(20),
                  columns = ['Python', 'Tensorflow', 'Keras'])
display(df)
#计算协方差
display(df.cov()) #两个属性之间的协方差
display(df['Python'].cov(df['Keras'])) #Python和Keras的协方差
#计算线性相关系数
display(df.corr())
display(df.corrwith(df['Python']))
```

协方差:
 $\text{Cov}(X, Y) = \frac{\sum_{i=1}^{n}(X_i - \bar{X})(Y_i - \bar{Y})}{n-1}$，
相关性系数:
 $r(X, Y) = \frac{\text{Cov}(X, Y)}{\sqrt{\text{Var}[X]\text{Var}[Y]}}$。

```python
#分箱操作(也叫面元划分,离散化)分为等距分箱和等频分箱
df = pd.DataFrame(data = np.random.randint(0, 150, size = (100, 3)), 
                  columns = ['Python', 'Tensorflow', 'Keras'])
display(df)
#等宽分箱
display(pd.cut(df.Python, bins = 3)) #把成绩等分成三份
#指定宽度分箱
display(pd.cut(df.Keras, #分箱数据
               bins = [0, 60, 90, 120, 150],  #分箱断点
               right = False,  #左闭右开(默认right = True,即左开右闭)
               labels = ['D', 'C', 'B', 'A']))
#等频分箱,使每组样本数大致相等
display(pd.qcut(df.Python, q = 4, 
               labels = ['D', 'C', 'B', 'A']))
```

```python
#分组聚合
#准备数据
df = pd.DataFrame(data = {'sex':np.random.randint(0, 2, size = 300), # 0男，1女
                          'class':np.random.randint(1, 9, size = 300),# 1~8八个班
                          'Python':np.random.randint(0, 151, size = 300),# Python成绩
                          'Keras':np.random.randint(0, 151, size =300),# Keras成绩
                          'Tensorflow':np.random.randint(0, 151, size = 300),
                          'Java':np.random.randint(0, 151,size = 300),
                          'C++':np.random.randint(0, 151, size = 300)})
df['sex'] = df['sex'].map({0:'男', 1:'女'})             # 将0，1映射成男女
display(df)
#分组
#根据性别分组，求得平均值，保留一位小数
display(df.groupby(by = 'sex').mean().round(1))
#统计男女数量
display(df.groupby(by = 'sex').size())
#根据性别和班级两个属性分组统计数量
display(df.groupby(by = ['sex', 'class']).size())
#多层索引
display(df.groupby(by = ['class', 'sex'])[['Python', 'Java']].max())
#数据重塑
display(df.groupby(by = ['class', 'sex'])[['Python', 'Java']].max().unstack)
```

```python
#分组聚合apply,transform
#apply返回汇总后的情况，对于每一个分组大类都只返回一个结果
display(df.groupby(by = ['class', 'sex'])[['Python', 'Java']].apply(np.mean).round(1))
#transform返回所有元素
display(df.groupby(by = ['class','sex'])[['Python','Keras']].transform(np.mean).round(1))
```

```python
#分组聚合agg
#按照班级和性别进行划分,统计 Tensorflow 和 Keras 这两门学科的最大值,最小值,个数
display(df.groupby(by = ['class', 'sex'])[['Tensorflow', 'Keras']].agg(
    [np.max, np.min, pd.Series.count]
))
```
