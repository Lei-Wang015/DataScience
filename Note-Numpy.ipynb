```python
#引入Numpy
import numpy as np
```

```python
#数组初始化
a = np.array([1, 2, 3, 4, 5])
print(a)
```

```python
#创建包含连续元素的数组
a = np.arange(1, 13)
print(a)
```

```python
#创建元素全部为0的数组
a = np.zeros(6)
b = np.zeros(shape = 6)
print(a)
print(b)
```

```python
#创建元素全部为1的数组
a = np.ones(10)
b = np.ones(shape = 10)
print(a)
print(b)
```

```python
#创建用特定值填充的数组
a = np.full(6, 666)
b = np.full(shape = 6, fill_value = 666)
print(a)
print(b)
```

```python
#在0~100生成10个随机整数
a = np.random.randint(0, 100, size = 10)
print(a)
#在0~10内随机生成三行五列二维数组
a = np.random.randint(0, 10, size = (3, 5))
print(a)
#三维数组
a = np.random.randint(0, 10, size = (2, 4, 6))
print(a)
```

```python
#随机生成10个正态分布数，平均值是0， 方差是1
a = np.random.randn(10)
print(a)
#随机生成10个正态分布数，平均值是5，标准差为2
a = np.random.randn(10) * 2 + 5
print(a)
```

```python
#生成等差数列,把1到100线性等间距分成19份
a = np.linspace(1, 100, 19)
print(a)
```

```python
#生成等比数列，把2^0到2^10对数等间距分为11份
#base表示底数，0 10表示指数，输出形式为科学计数法
b = np.logspace(0, 10, base = 2, num = 11)
print(b)
```

```python
#np.set_printoptions(suppress=True)将输出显示为小数或整数
np.set_printoptions(suppress=True)
b = np.logspace(0, 10, base = 2, num = 11)
print(b)
```

```python
#数组的维度
a = np.random.randint(0, 10, size=(2, 3, 5))
print(a.ndim)
```

```python
#数组的形状
a = np.random.randint(0, 10, size=(2, 3, 5))
print(a.shape)
```

```python
#数组元素的总数
a = np.random.randint(0, 10, size=(2, 3, 5))
print(a.size)
```

```python
#数组元素的数据类型
a = np.random.randint(0, 10, size=(2, 3, 5))
print(a.dtype)
```

```python
#数组中每个元素的大小（以字节为单位）
a = np.random.randint(0, 10, size=(2, 3, 5))
print(a.itemsize)
```

```python
#数组变形
a = np.arange(1, 13).reshape(3, 4)
print(a)
display(a) #更美观地打印数组
b = np.arange(12).reshape(-1, 6) #-1便是的是最后计算 x * 6 = 3 * 4 -> x = 2
c = np.arange(12).reshape(2, 6)
d = a.reshape(-1) # x = 3 * 4
print(b)
print(c)
print(d)
```

```python
#数组运算 + - * /
a = np.arange(12).reshape(3, 4)
b = np.full(12, 2).reshape(3, 4)
print(a)
print(b)
print(a + b)
print(a + 2) #每个元素都加2
print(a - b)
print(a - 2)
print(a * b)
print(a * 2)
print(a / b)
print(a / 2)
```

```python
#数组次方
a = np.arange(1, 7).reshape(2, 3)
print(np.power(a, 3))
print(a ** 2)
```

```python
#数组的取余和地板除法
a = np.arange(12).reshape(3, 4)
b = np.full(12, 2).reshape(3, 4)
print(a % b)
print(a // b)
```

```python
#数组的逻辑运算（返回bool）
a = np.arange(1, 13).reshape(3, 4)
b = np.full(12, 2).reshape(3, 4)
print(a > 5)
print(a == 5)
print(a > b)
print(a == b)
print(a[(a > 5) & (a < 12)]) #一定要把两个条件加括号不然会报错
print(a[(a < 5) | (a > 7)])
```

```python
#+= -= *= //=操作
a = np.arange(12).reshape(3, 4)
a += 10
print(a)
a -= 10
print(a)
a *= 10
print(a)
a //= 10 #注意是地板除法
print(a)
```

```python
#索引与切片
a = np.arange(12).reshape(3, 4)
print(a)
print(a[0]) #第0行
print(a[-1]) #最后一行
print(a[[0, 2]]) #第零行和第二行
print(a[0][1 : 3]) 
print(a[0 : 2][0]) #相当于先新建数组b = [0, 1, 2, 3][4, 5, 6, 7],后索引b[0]
print(a[:])
print(a[:: - 1])
```

```python
#数组堆叠
a = np.arange(12).reshape(3, 4) #默认axis = 0，代表行操作，列不变
b = np.full(12, 2).reshape(3, 4)
display(np.concatenate([a, b]))
display(np.concatenate([a, b, a]))
b = np.ones((3, 5))
display(np.concatenate([a, b], axis = 1)) #代表列操作，行不变
```

```python
#数组拆分
a = np.arange(18).reshape(3, 6)
display(np.split(a, 2, axis = 1)) #列操作，均分为三组
a = np.arange(18).reshape(3, 6)
display(np.split(a, [0, 1, 3], axis = 1)) #在第0，1，3列节点分隔 [0,1) [1,3) [3, 6)
```

```python
#数组转置
a = np.random.randint(0, 100, size = (3, 4, 5))
display(a)
display(a.T) #默认把维度0，1，2转为2，1，0
display(np.transpose(a, axes = [1, 0, 2])) #交换第1维和第2维
```

```python
#广播机制
#一维数组的广播
a = np.arange(1, 13).reshape(3, 4) #3, 4
b = np.arange(4)  #相当于1, 4
print(a + b) #为实现相加b会从最后一维度向前自动广播到3，4以达到目标形式
#二维数组的广播
a = np.random.randint(0, 10, size = (4, 5)) #4, 5
b = a.mean(axis = 1)  #1, 4 最后一个维度不相等还都不是1，导致形式无法统一，相加会报错
print(b)
b = b.reshape(4, 1) #手动统一形状
print(a + b)
#三维数组的广播
a = np.array([0, 1, 2, 3, 4, 5, 6, 7] * 3).reshape(3, 4, 2)
b = np.array([0, 1, 2, 3, 4, 5, 6, 7]).reshape(4, 2) #相当于1, 4, 2
print(a + b)
```

```python
#数学相关函数-一元元素级函数
display(np.pi) #圆周率
display(np.sin(np.pi / 2)) #计算正弦值，要用弧度制
display(np.cos(np.pi / 2)) #计算过程中精度丢失，就当作0看
display(np.cos(np.pi / 2).round(5)) #保留五位小数
display(np.round(np.exp(1), 2)) #保留小数到指定位
display(np.ceil(2.5)) #向上取整
display(np.floor(2.5)) #向下取整
display(np.exp(1)) #e的指数函数
display(np.sqrt(1024)) #开平方
display(np.square(8)) #平方
display(np.power(2, 3)) #幂运算
display(np.log2(16)) #计算log以2为底16的对数，numpy仅内置了np.log(底e),log10,log2,其他的用换底公式
```

```python
#数学相关函数-二元元素级函数
a = np.array([6, 6, 0, 7, 2, 5])
b = np.array([9, 5, 6, 3, 4, 2])
display(np.maximum(a, b)) #比较后返回两个数组对应位置的最大值
display(np.minimum(a, b)) #返回最小值
```

```python
#数学相关函数-逻辑与统计函数
a = np.arange(12).reshape(3, 4)
print(a)
print(np.all(a)) #判断是否所有元素为真
print(np.any(a)) #判断是否至少有一个元素为真
print(np.trace(a)) #计算矩阵的迹,对于非方阵会计算行标等于列标的元素和
```

```python
#计算点积(内积)
a = np.random.randint(0, 10, size = (2, 2))
display(a)
display(np.inner(a[0], a))
```

```python
#元素裁剪
a = np.arange(12).reshape(3, 4)
display(a)
display(np.clip(a, 3, 6)) #把小于3的都变成3，大于6的都变成6
```

```python
#where函数
a = np.arange(12).reshape(3, 4)
print(np.where(a > 6)) #返回索引 即(1, 3)(2, 0)(2, 1)(2, 2) (2, 3)
print(a[np.where(a > 6)]) #返回值
print(np.where(a > 6, 10, a)) #np.where(condition, x, y)满足条件时取x否则取y 相当于condition? x : y
```

```python
#排序-np.sort()
a = np.random.randint(0, 100, size = (3, 4))
print(a)
display(np.sort(a)) #deep copy默认操作行，即axis = 0
display(a)
```

```python
#排序-arr.argsort()
a = np.random.randint(0, 10, size = (3, 4))
display(a)
display(np.argsort(a)) #返回排序索引,也是deep copy
display(a)
```

```python
#排序-arr.sort()
a = np.random.randint(0, 10, size = (3, 4))
display(a)
a.sort()
display(a) #改变原数组，本质是shallow copy
```

```python
#集合运算函数-只使用一维(1d)数组
a = np.arange(12)
b = np.arange(4, 16)
display(np.intersect1d(a, b)) #交集
display(np.union1d(a, b)) #并集
display(np.setdiff1d(a, b)) #差集
```

```python
#数学与统计函数
a = np.arange(1, 13).reshape(3, 4)
display(np.min(a))
display(np.max(a, axis = 1))
display(np.mean(a)) #平均值
display(np.median(a)) #中位数
display(np.sum(a)) #所有元素和
display(np.sum(a, axis = 0)) #列和
display(np.sum(a, axis = 1)) #行和
display(np.std(a)) #标准差
display(np.var(a)) #方差
display(np.cumsum(a)) #累计和
display(np.cumprod(a)) #累计积
display(np.argmin(a)) #返回最小值索引
display(np.argmax(a)) #返回最大值索引
display(np.argwhere(a > 6)) 
display(np.cov(a)) #计算协方差矩阵
display(np.corrcoef(a)) #计算相关系数矩阵
```

```python
#矩阵运算
#矩阵的数量积
a = np.arange(6).reshape(2, 3)
b = np.full(6, 2).reshape(3, 2)
print(np.dot(a, b)) #三种形式
print(a @ b)
print(a.dot(b))
```

```python
#矩阵运算
#逆矩阵
from numpy.linalg import inv, det, eig, qr, svd
A = np.random.randint(1, 10, size = (3, 3))
display(det(A)) #行列式
B = inv(A) #逆矩阵
display(B)
display(A.dot(B))
#inv:inverse
#det:determinant
#eig:eigenvalues
#qr:QR decomposition
#svd:singular value decomposition
```
